# Data Ingestion from Preprocessed CSV Files

This notebook loads preprocessed Synthea data from CSV files.

```python
import sys
sys.path.append('..')

import pandas as pd
import numpy as np
from pathlib import Path

from src.data.csv_loader import SyntheaDataLoader
from src.utils.logging_config import setup_logging
from src.utils.reproducibility import set_random_seeds
```

```python
# Setup
setup_logging()
set_random_seeds(42)

# Initialize data loader
data_loader = SyntheaDataLoader(data_dir='../data/raw/preprocessing')
```

## 1. Load Clinical Data

```python
clinical_data = data_loader.load_clinical_data()
print(f"Loaded {len(clinical_data)} clinical records")
print(f"Features: {len(clinical_data.columns)}")
print(f"Unique patients: {clinical_data['patientID'].nunique()}")
clinical_data.head()
```

## 2. Load Genomic Data

```python
genomic_data = data_loader.load_genomic_data()
print(f"Loaded {len(genomic_data)} genomic records")
print(f"Features: {len(genomic_data.columns)}")
print(f"Unique patients: {genomic_data['patientID'].nunique()}")
genomic_data.head()
```

## 3. Load Imaging Features

```python
imaging_data = data_loader.load_imaging_data()
print(f"Loaded {len(imaging_data)} imaging records")
print(f"Features: {len(imaging_data.columns)}")
print(f"Unique patients: {imaging_data['patientID'].nunique()}")
imaging_data.head()
```

## 4. Data Summary

```python
print("\n" + "="*80)
print("DATA SUMMARY")
print("="*80)
print(f"Clinical records: {len(clinical_data):,}")
print(f"Genomic records: {len(genomic_data):,}")
print(f"Imaging records: {len(imaging_data):,}")
print(f"\nTotal features: {len(clinical_data.columns) + len(genomic_data.columns) + len(imaging_data.columns):,}")
print("="*80)
```

## 5. Save Raw Data for Preprocessing

Save data to data/raw directory for the preprocessing notebook.

```python
# Save to data/raw for preprocessing notebook
data_dir = Path('../data/raw')
data_dir.mkdir(parents=True, exist_ok=True)
clinical_data.to_csv(data_dir / 'clinical_raw.csv', index=False)
genomic_data.to_csv(data_dir / 'genomic_raw.csv', index=False)
imaging_data.to_csv(data_dir / 'imaging_raw.csv', index=False)
print(f"✓ Saved clinical_raw.csv ({len(clinical_data)} records)")
print(f"✓ Saved genomic_raw.csv ({len(genomic_data)} records)")
print(f"✓ Saved imaging_raw.csv ({len(imaging_data)} records)")
print("\nData ready for preprocessing!")
```
